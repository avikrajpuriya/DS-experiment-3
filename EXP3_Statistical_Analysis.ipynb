# Experiment 3 – Measuring Central Tendency and Variability

**Aim:** To calculate mean, median, mode, range, variance, standard deviation and IQR for single and grouped data using Python.

**Dataset:** `urban_climate.csv`

Keep the CSV file and this notebook in the same folder. Run the cells from top to bottom.

## 1. Read the dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("urban_climate.csv")
print("Number of records:", len(df))
df.head()
```

**Inference:** The dataset is loaded successfully. It contains climate records for different cities.

## 2. Frequency, mean, median and mode

```python
# Frequency of city records
print(df["city"].value_counts())

# Mean, median and mode for three numerical columns
columns = ["temperature_celsius", "humidity_percent", "precipitation_mm"]

for col in columns:
    print("\n", col)
    print("Frequency:", df[col].count())
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode().tolist())
```

**Inference:** Mean gives the average, median gives the middle value and mode gives the most repeated value. Frequency shows the number of available records.

## 3. Manual calculations without statistical inbuilt functions
The following functions use loops and simple calculations. Bubble sort is used to arrange values.

```python
# Helper functions for calculating statistics manually.

def manual_sort(values):
    # Merge sort is used to arrange values without using sorted().
    if len(values) <= 1:
        return values[:]

    middle = len(values) // 2
    left = manual_sort(values[:middle])
    right = manual_sort(values[middle:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result


def get_mean(values):
    total = 0
    for value in values:
        total += value
    return total / len(values)


def get_median(values):
    data = manual_sort(values)
    n = len(data)
    if n % 2 == 0:
        return (data[n // 2 - 1] + data[n // 2]) / 2
    return data[n // 2]


def get_mode(values):
    frequency = {}
    for value in values:
        frequency[value] = frequency.get(value, 0) + 1

    highest = max(frequency.values())
    if highest == 1:
        return "No mode"

    modes = []
    for value in frequency:
        if frequency[value] == highest:
            modes.append(value)
    return modes


def get_variance(values):
    mean = get_mean(values)
    squared_total = 0
    for value in values:
        squared_total += (value - mean) ** 2
    if len(values) < 2:
        return 0
    return squared_total / (len(values) - 1)


def get_standard_deviation(values):
    return get_variance(values) ** 0.5


def get_quartiles(values):
    data = manual_sort(values)
    n = len(data)
    q2 = get_median(data)

    if n % 2 == 0:
        lower = data[:n // 2]
        upper = data[n // 2:]
    else:
        lower = data[:n // 2]
        upper = data[n // 2 + 1:]

    q1 = get_median(lower)
    q3 = get_median(upper)
    return q1, q2, q3, q3 - q1


def read_numeric_column(rows, column):
    values = []
    for row in rows:
        try:
            values.append(float(row[column]))
        except (ValueError, TypeError, KeyError):
            pass
    return values
```

```python
# Use the manual functions on the temperature column
values = df["temperature_celsius"].dropna().tolist()

print("Mean:", get_mean(values))
print("Median:", get_median(values))
print("Mode:", get_mode(values))
print("Sample variance:", get_variance(values))
print("Sample standard deviation:", get_standard_deviation(values))
print("Q1, Q2, Q3 and IQR:", get_quartiles(values))
```

**Inference:** The statistics are calculated manually using loops and formulas rather than ready-made statistical functions.

## 4. Variance and standard deviation for multiple attributes

```python
for col in ["temperature_celsius", "humidity_percent", "precipitation_mm"]:
    values = df[col].dropna().tolist()
    print("\nAttribute:", col)
    print("Variance:", get_variance(values))
    print("Standard deviation:", get_standard_deviation(values))
```

**Inference:** Variance and standard deviation describe how far the values are spread around the mean. A larger value indicates greater variation.

## 5. Quartiles and IQR for multiple attributes

```python
for col in ["temperature_celsius", "humidity_percent", "precipitation_mm"]:
    values = df[col].dropna().tolist()
    q1, q2, q3, iqr = get_quartiles(values)
    print("\nAttribute:", col)
    print("Q1:", q1)
    print("Median (Q2):", q2)
    print("Q3:", q3)
    print("IQR:", iqr)
```

**Inference:** IQR is the difference between Q3 and Q1. It describes the spread of the middle 50% of the data.

## 6. Grouped data analysis by city

```python
for city, group in df.groupby("city"):
    print("\nCity:", city)
    print("Frequency:", len(group))

    for col in ["temperature_celsius", "humidity_percent", "precipitation_mm"]:
        values = group[col].dropna().tolist()
        q1, q2, q3, iqr = get_quartiles(values)

        print("\n", col)
        print("Mean:", get_mean(values))
        print("Mode:", get_mode(values))
        print("Median:", get_median(values))
        print("Variance:", get_variance(values))
        print("Standard deviation:", get_standard_deviation(values))
        print("IQR:", iqr)
```

**Inference:** Grouping by city allows us to compare the central tendency and spread of climate measurements for each city.

## Note on calculations

- Variance and standard deviation in the Python programs use the **sample formula** with `n - 1`.
- Quartiles are calculated by dividing the sorted values into lower and upper halves and finding their medians.
- Small differences may occur if another software uses a different quartile calculation method.
